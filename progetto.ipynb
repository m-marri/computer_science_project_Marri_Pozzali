```python
import pandas as pd
```

```python
CA_videos = pd.read_csv("videos/CAvideos.csv.zst")
DE_videos = pd.read_csv("videos/DEvideos.csv.zst")
FR_videos = pd.read_csv("videos/FRvideos.csv.zst")
IN_videos = pd.read_csv("videos/INvideos.csv.zst")
US_videos = pd.read_csv("videos/USvideos.csv.zst")
GB_videos = pd.read_csv("videos/GBvideos.csv.zst")
JP_videos = pd.read_csv("videos/JPvideos.csv.zst", encoding='utf-8', encoding_errors= 'replace')
MX_videos = pd.read_csv("videos/MXvideos.csv.zst", encoding='utf-8', encoding_errors= 'replace')
RU_videos = pd.read_csv("videos/RUvideos.csv.zst", encoding='utf-8', encoding_errors= 'replace')
KR_videos = pd.read_csv("videos/KRvideos.csv.zst", encoding='utf-8', encoding_errors= 'replace')
```

### 1. Create a single dataframe with the concatenation of all input csv files, adding a column called country

```python
CA_videos['country']='CA'
DE_videos['country']='DE'
FR_videos['country']='FR'
GB_videos['country']='GB'
IN_videos['country']='IN'
JP_videos['country']='JP'
KR_videos['country']='KR'
MX_videos['country']='MX'
RU_videos['country']='RU'
US_videos['country']='US'
lista= (CA_videos, DE_videos, FR_videos, GB_videos, IN_videos, JP_videos, KR_videos, MX_videos, RU_videos, US_videos)
unione= pd.concat(lista, ignore_index=True)
unione
```

### 2. Extract all videos that have no tag

```python
no_tags = unione[unione["tags"] == "[none]"]
no_tags
```

### 3. For each channel, determine the total number of views

```python
tot_views = unione.groupby("channel_title")[["views"]].sum()
tot_views
```

### 4. Save all rows with disabled comments and disabled ratings, or that have video_error_or_removed in a new dataframe called excluded, and remove those rows from the original dataframe

```python
excluded = unione[((unione["comments_disabled"] == True) &
    (unione["ratings_disabled"] == True)) |
    (unione["video_error_or_removed"] == True)
    ]
unione_excluded = unione.drop(excluded.index)
unione_excluded
```

### 5. Add a like_ratio column storing the ratio between the number of likes and of dislikes

```python
unione_excluded["like_ratio"] = unione_excluded["likes"]/unione_excluded["dislikes"]
```

### 6 Cluster the publish time into 10-minute intervals (e.g. from 02:20 to 02:30)

```python
unione_excluded["publish_time"] = pd.to_datetime(unione_excluded["publish_time"])
unione_excluded["range_10min"] = unione_excluded["publish_time"].dt.floor("10min").dt.time
unione_excluded.head()
```

### 7 For each interval, determine the number of videos, average number of likes and of dislikes.

```python
union_cluster= unione_excluded.groupby("range_10min").agg({"video_id": "count" ,
                                                  "likes": "mean",
                                                  "dislikes": "mean"})
union_cluster
```

### 8. For each tag, determine the number of videos

```python
no_tags2 = unione_excluded[unione_excluded["tags"] == "[none]"]
unione_notags = unione_excluded.drop(no_tags2.index)
tags = unione_notags["tags"].str.replace('"', '').str.split(r"[|/]")
tags2= tags.explode()
tot_tags = tags2.value_counts()
tot_tags
```

### 9. Find the tags with the largest number of videos

```python
tot_tags.head(10)
```

### 10. For each (tag,country) pair, compute average ratio like/dislikes

```python
unione_tags = (unione_notags.assign(tags = unione_notags["tags"].str.replace('"','').str.split(r"[|/]")).explode("tags").reset_index(drop=True))
unione_tags.groupby(['country','tags'])['like_ratio'].mean()
unione_tags
```

### 11. For each (trending_date, country) pair, the video with the largest number of views

```python
id_views = unione_tags.groupby(['trending_date','country'])['views'].idxmax()
id_views = unione_tags.loc[id_views]
id_views[['trending_date','country','title']]
```

### 12. Divide trending_date into three columns: year, month, day

```python
unione_excluded['day'] = unione_excluded['publish_time'].dt.day
unione_excluded['month'] = unione_excluded['publish_time'].dt.month
unione_excluded['year'] = unione_excluded['publish_time'].dt.year
```

### 13. For each (month, country) pair, the video with the largest number of views

```python
id_views2 = unione_excluded.groupby(['month','country'])['views'].idxmax()
id_views2 = unione_excluded.loc[id_views2]
id_views2[['month','country','title']]
```

### 14. Read all json files with the video categories

```python
CA_category = pd.read_json("CA_category_id.json")
DE_category = pd.read_json("DE_category_id.json")
FR_category = pd.read_json("FR_category_id.json")
IN_category = pd.read_json("IN_category_id.json")
US_category = pd.read_json("US_category_id.json")
GB_category = pd.read_json("GB_category_id.json")
JP_category = pd.read_json("JP_category_id.json")
MX_category = pd.read_json("MX_category_id.json")
RU_category = pd.read_json("RU_category_id.json")
KR_category = pd.read_json("KR_category_id.json")

lista_category = (CA_category, DE_category, FR_category, GB_category, IN_category, JP_category, KR_category, MX_category, RU_category, US_category)
unione_category = pd.concat(lista_category, ignore_index=True)
unione_category
```

```python
unione_category_normalized = pd.json_normalize(unione_category['items'])

unione_category_normalized = unione_category_normalized.rename(columns={'snippet.channelId': 'channelId',
                                                                        'snippet.title': 'title',
                                                                        'snippet.assignable': 'assignable'})

unione_category_normalized
```

### 15. For each country, determine how many videos have a category that is not assignable.

```python
unione_category_normalized['id'] = unione_category_normalized['id'].astype(int)
unione_join = pd.merge(unione_excluded, unione_category_normalized, left_on = 'category_id', right_on = 'id', how='left')
unione_join_false = unione_join[unione_join['assignable'] == False]
unione_join_false.groupby('country')['video_id'].count()
```
